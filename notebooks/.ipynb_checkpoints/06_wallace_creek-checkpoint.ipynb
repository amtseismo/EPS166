The creek, named after USGS geologist Robert Wallace, crosses the central San Andreas fault in the Carrizo Plain of central California.
```python
# ── Download and crop the GeoTIFF DEM ────────────────────────────────────────
DATA_URL = (
    "https://raw.githubusercontent.com/amtseismo/EPS166/"
    "b37ac7a2df7d665bee1d1a4b0e269aa061d8109e/datasets/WallaceCreek_DEM.tif"
)

# Crop window matching the original Wallace Creek site extent
CROP = dict(left=242250, right=243220, bottom=3906400, top=3907380)  # UTM Zone 11

print(f"Downloading DEM from:\n  {DATA_URL}")
with urlopen(DATA_URL, timeout=120) as r:
    tif_bytes = r.read()
print(f"Downloaded {len(tif_bytes)/1e6:.1f} MB — reading crop window...")

with rasterio.open(io.BytesIO(tif_bytes)) as src:
    window  = from_bounds(**CROP, transform=src.transform)
    dem     = src.read(1, window=window).astype(np.float32)
    NODATA  = src.nodata
    transform = src.window_transform(window)

CELL  = transform.a                         # pixel size in metres (1.0 m)
NCOLS, NROWS = dem.shape[1], dem.shape[0]
XLL   = transform.c                         # Easting of left edge
YLL   = transform.f - NROWS * CELL          # Northing of bottom edge

dem[dem == NODATA] = np.nan

# Coordinate arrays (cell centres)
east  = XLL + (np.arange(NCOLS) + 0.5) * CELL
north = YLL + (NROWS - np.arange(NROWS) - 0.5) * CELL

EXT = [XLL, XLL + NCOLS*CELL, YLL, YLL + NROWS*CELL]

print(f"DEM: {NCOLS} x {NROWS} cells at {CELL:.1f} m resolution")
print(f"Easting:  {XLL:.0f} – {XLL+NCOLS*CELL:.0f} m")
print(f"Northing: {YLL:.0f} – {YLL+NROWS*CELL:.0f} m  (UTM Zone 11)")
print(f"Elevation: {np.nanmin(dem):.1f} – {np.nanmax(dem):.1f} m")
```

## Part I: Load and visualize the DEM

We compute a hillshade — a synthetic image of the terrain illuminated from a given sun direction — to reveal subtle topographic features that are hard to see in raw elevation data.


```python
def hillshade(dem, cell, azimuth_deg=315, altitude_deg=45):
    """Compute a greyscale hillshade from a DEM.
    azimuth_deg : sun direction (clockwise from N, default NW = 315°)
    altitude_deg: sun angle above horizon (default 45°)
    """
    dx, dy = np.gradient(dem, cell, cell)
    slope  = np.arctan(np.sqrt(dx**2 + dy**2))
    aspect = np.arctan2(-dy, dx)               # counter-clockwise from E
    az  = np.deg2rad(azimuth_deg)
    alt = np.deg2rad(altitude_deg)
    hs  = (np.cos(alt) * np.cos(slope)
           + np.sin(alt) * np.sin(slope) * np.cos(az - aspect))
    return np.clip(hs, 0, 1)

hs = hillshade(dem, CELL, azimuth_deg=315, altitude_deg=45)

# Extent for imshow: [x_left, x_right, y_bottom, y_top]
EXT = [XLL, XLL + NCOLS*CELL, YLL, YLL + NROWS*CELL]

fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(hs, cmap="gray", extent=EXT, origin="upper", vmin=0, vmax=1)
ax.set_xlabel("Easting (m, UTM Zone 11)", fontsize=11)
ax.set_ylabel("Northing (m)", fontsize=11)
ax.set_title("Wallace Creek LiDAR — hillshade (315°/45°)", fontsize=12)

# Scale bar
ax.plot([XLL+50, XLL+150], [YLL+30, YLL+30], "k-", lw=3)
ax.text(XLL+100, YLL+50, "100 m", ha="center", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# ── Slope map ─────────────────────────────────────────────────────────────────
dx, dy = np.gradient(dem, CELL, CELL)
slope_deg = np.rad2deg(np.arctan(np.sqrt(dx**2 + dy**2)))

fig, axes = plt.subplots(1, 2, figsize=(14, 7), sharey=True)
axes[0].imshow(hs, cmap="gray", extent=EXT, origin="upper")
axes[0].set_title("Hillshade (315°/45°)", fontsize=11)

im = axes[1].imshow(slope_deg, cmap="hot_r", extent=EXT, origin="upper", vmin=0, vmax=30)
axes[1].set_title("Slope (degrees)", fontsize=11)
plt.colorbar(im, ax=axes[1], label="slope (°)", shrink=0.7)

for ax in axes:
    ax.set_xlabel("Easting (m)", fontsize=10)
axes[0].set_ylabel("Northing (m)", fontsize=10)
plt.tight_layout()
plt.show()
```

## Part II: Define the fault coordinate system

To measure fault-parallel offsets we rotate from UTM (Easting, Northing) coordinates into a fault-aligned system where:

- $x_\parallel$ = distance along the fault (positive NW)
- $x_\perp$ = distance perpendicular to the fault (positive NE, toward the hills)

The San Andreas at Wallace Creek strikes approximately **N45°W (315°)**.


```python
# ── Fault geometry ────────────────────────────────────────────────────────────
# Adjust these three values to align the trace with the LiDAR scarp.
FAULT_STRIKE_DEG = 320       # azimuth clockwise from N (e.g. 315 = N45°W)
FAULT_ORIGIN_E   = 242662.0    # UTM Easting of a point on the fault trace (m)
FAULT_ORIGIN_N   = 3906992.0   # UTM Northing

# Unit vectors derived from strike — do not edit below this line
strike_rad = np.deg2rad(FAULT_STRIKE_DEG)
s_hat = np.array([ np.sin(strike_rad),  np.cos(strike_rad)])  # along fault (strike direction)
n_hat = np.array([ np.cos(strike_rad), -np.sin(strike_rad)])  # perpendicular (NE, toward hills)

print(f"Fault strike:               {FAULT_STRIKE_DEG}°")
print(f"Along-fault unit vector:    {s_hat.round(3)}")
print(f"Fault-normal unit vector:   {n_hat.round(3)}")

def utm_to_fault(E, N):
    """Convert UTM (E, N) to fault-parallel / fault-perpendicular coordinates.
    x_parallel > 0 is in the strike direction.
    x_perp     > 0 is NE (toward the Temblor Range).
    """
    dE = E - FAULT_ORIGIN_E
    dN = N - FAULT_ORIGIN_N
    return dE*s_hat[0] + dN*s_hat[1], dE*n_hat[0] + dN*n_hat[1]
```

```python
# ── Annotated overview map ────────────────────────────────────────────────────
# Draw the fault trace across the DEM
t_vals   = np.linspace(-500, 500, 100)  # metres along fault
fault_E  = FAULT_ORIGIN_E + t_vals * s_hat[0]
fault_N  = FAULT_ORIGIN_N + t_vals * s_hat[1]

fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(hs, cmap="gray", extent=EXT, origin="upper", vmin=0, vmax=1)
ax.plot(fault_E, fault_N, "r-", lw=1.5, alpha=0.8, label="SAF trace (approx.)")

# Mark the fault origin point
ax.plot(FAULT_ORIGIN_E, FAULT_ORIGIN_N, "r+", ms=14, mew=2)

# Compass
ax.annotate("", xy=(XLL+870, YLL+120), xytext=(XLL+870, YLL+40),
            arrowprops=dict(arrowstyle="->", color="k", lw=2))
ax.text(XLL+870, YLL+130, "N", ha="center", fontsize=12, fontweight="bold")

# Scale bar
ax.plot([XLL+50, XLL+150], [YLL+30, YLL+30], "k-", lw=3)
ax.text(XLL+100, YLL+50, "100 m", ha="center", fontsize=9)

ax.set_xlabel("Easting (m, UTM Zone 11)", fontsize=11)
ax.set_ylabel("Northing (m)", fontsize=11)
ax.set_title("Wallace Creek — fault trace and channel system", fontsize=12)
ax.legend(frameon=False, fontsize=10)
ax.set_xlim(EXT[0], EXT[1])
ax.set_ylim(EXT[2], EXT[3])
plt.show()
```

## Part III: Measure channel offsets

### How the measurement works

The San Andreas fault is right-lateral: the SW block (Pacific plate) has moved NW relative to the NE block (Temblor Range). Each time Wallace Creek carved a new channel, fault slip began offsetting it. The **offset** of each channel is how far it has been displaced along the fault since it first formed.

We measure this using **fault-parallel profiles** — elevation transects drawn parallel to the fault, one on each side:

- **NE profile** (+30 m from fault): samples the channel on the stable, undeflected side. Channels appear as **dips** (valleys) in the elevation. Their fault-parallel positions are approximately known from the hillshade map.
- **SW profile** (−30 m from fault): samples the same channels on the offset side. Each channel has been shifted **NW** (toward positive fault-parallel values) by dextral slip.

**Offset = SW position − NE position** (positive = NW displacement = right-lateral).

```python
def extract_profile(dem_grid, east_arr, north_arr, E0, N0, along_hat, n_points=400, half_width_m=600):
    """Extract a swath profile along a line through (E0, N0) in direction along_hat.
    Returns (along_distance_m, median_elevation, std_elevation) arrays.
    """
    # Sample points along the profile
    t = np.linspace(-half_width_m, half_width_m, n_points)
    E_pts = E0 + t * along_hat[0]
    N_pts = N0 + t * along_hat[1]

    # Bilinear interpolation
    dE = east_arr[1] - east_arr[0]
    dN = north_arr[0] - north_arr[1]  # north_arr is decreasing with row index

    col_f = (E_pts - east_arr[0]) / dE
    row_f = (north_arr[0] - N_pts) / dN

    col_i = np.clip(col_f.astype(int), 0, dem_grid.shape[1]-2)
    row_i = np.clip(row_f.astype(int), 0, dem_grid.shape[0]-2)

    fc = col_f - col_i
    fr = row_f - row_i

    elev = (
        dem_grid[row_i,   col_i  ] * (1-fc) * (1-fr)
      + dem_grid[row_i,   col_i+1] *  fc    * (1-fr)
      + dem_grid[row_i+1, col_i  ] * (1-fc) *  fr
      + dem_grid[row_i+1, col_i+1] *  fc    *  fr
    )
    elev[elev == NODATA] = np.nan
    return t, elev
```

```python
# ── Fault-parallel profiles ────────────────────────────────────────────────────
PERP_OFFSET = 30.0   # metres each side of fault

NE_origin = (FAULT_ORIGIN_E + PERP_OFFSET * n_hat[0],
             FAULT_ORIGIN_N + PERP_OFFSET * n_hat[1])
SW_origin = (FAULT_ORIGIN_E - PERP_OFFSET * n_hat[0],
             FAULT_ORIGIN_N - PERP_OFFSET * n_hat[1])

t_NE, elev_NE = extract_profile(dem, east, north, *NE_origin, s_hat, half_width_m=450)
t_SW, elev_SW = extract_profile(dem, east, north, *SW_origin, s_hat, half_width_m=450)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ── Overview map showing profile locations ────────────────────────────────────
ax = axes[0]
ax.imshow(hs, cmap="gray", extent=EXT, origin="upper", vmin=0, vmax=1)
ax.plot(fault_E, fault_N, "r-", lw=1.5, alpha=0.8, label="fault trace")
t_line = np.linspace(-450, 450, 200)
ax.plot(NE_origin[0] + t_line*s_hat[0], NE_origin[1] + t_line*s_hat[1],
        "b--", lw=1.2, label="NE profile (+30 m)")
ax.plot(SW_origin[0] + t_line*s_hat[0], SW_origin[1] + t_line*s_hat[1],
        "g--", lw=1.2, label="SW profile (−30 m)")
ax.set_xlim(EXT[0], EXT[1]); ax.set_ylim(EXT[2], EXT[3])
ax.set_xlabel("Easting (m)"); ax.set_ylabel("Northing (m)")
ax.set_title("Profile locations", fontsize=11)
ax.legend(fontsize=8, frameon=False)

# ── NE and SW elevation profiles ──────────────────────────────────────────────
for ax, (t, elev, side, col) in zip(axes[1:], [
        (t_NE, elev_NE, "NE side (undeflected)", "b"),
        (t_SW, elev_SW, "SW side (offset)",       "g"),
]):
    ax.plot(t, elev, color=col, lw=1.5)
    ax.set_xlabel("Fault-parallel distance (m)", fontsize=10)
    ax.set_ylabel("Elevation (m)", fontsize=10)
    ax.set_title(side, fontsize=11)
    ax.xaxis.set_minor_locator(plt.MultipleLocator(20))
    ax.yaxis.set_minor_locator(plt.MultipleLocator(1))
    ax.grid(True, which="major", alpha=0.4)
    ax.grid(True, which="minor", alpha=0.15)
    ax.axvline(0, color="0.5", lw=1.2, ls="--")
    ax.text(3, np.nanmax(elev), "fault\norigin", fontsize=8, color="0.5", va="top")
    ylo = np.nanmin(elev) + 0.04 * (np.nanmax(elev) - np.nanmin(elev))
    ax.text(t.min() + 10, ylo, "← SE", ha="left",  fontsize=9, color="0.45")
    ax.text(t.max() - 10, ylo, "NW →", ha="right", fontsize=9, color="0.45")

plt.suptitle("Fault-parallel elevation profiles — NE vs SW sides", fontsize=12)
plt.tight_layout()
plt.show()
```

## Part IV: Calculate slip rates

The slip rate is offset divided by age:

$$
\dot{s} = \frac{\text{offset}}{\text{age}}
$$

The ages below were determined by Sieh and Jahns (1984) using **radiocarbon (<sup>14</sup>C) dating** of organic material from sediment deposits in and around each channel. Radiocarbon dating measures the decay of <sup>14</sup>C in organic material to give an age with a ± uncertainty. The age used here is when each channel was active at the fault — i.e. when its offset was zero.

| Channel | Age (ka) | ±1σ (ka) | Dating constraint |
|---|---|---|---|
| Active | 3.7 | 0.2 | Top of Hh channel-fill deposits |
| 2nd abandoned | 13.3 | 1.6 | Base of Py fan complex |

Because both offset and age have uncertainty, the slip rate also has uncertainty. The cell below propagates these uncertainties using simple error propagation and also plots offset against age so you can see the relationship visually.


```python
# ── Ages from Sieh and Jahns (1984) ───────────────────────────────────────────
AGES_KA = {
    "Active":        (3.7,  0.2),   # (age in ka, 1-sigma uncertainty in ka)
    "2nd abandoned": (13.3, 1.6),
}

# ── Slip rate with simple error propagation ────────────────────────────────────
# For rate = offset / age, the fractional uncertainty adds in quadrature:
#   (σ_rate / rate)² = (σ_offset / offset)² + (σ_age / age)²

print(f"{'Channel':20s}  {'Slip rate (mm/yr)':>18}  {'Lit. (Sieh & Jahns)'}")
print("-" * 60)
SR = {}
for name in MY_OFFSETS:
    off, off_err = MY_OFFSETS[name]
    age_ka, age_err_ka = AGES_KA[name]
    age_yr  = age_ka  * 1e3
    rate    = off / age_yr * 1e3        # mm/yr  (off in m, age in yr → ×1000)
    rate_err = rate * np.sqrt((off_err/off)**2 + (age_err_ka/age_ka)**2)
    SR[name] = (rate, rate_err)
    print(f"{name:20s}  {rate:>8.1f} ± {rate_err:<6.1f} mm/yr  33.9 mm/yr")
```

```python
# ── Offset vs. age plot ────────────────────────────────────────────────────────
# Student measurements (Active + 2nd abandoned) plotted with error bars.
# The 1st abandoned channel is shown from Sieh & Jahns (1984) for reference
# — it is too subtle to measure reliably from the LiDAR alone.
COLS = {"Active": "#2a78d6", "2nd abandoned": "#eb6834"}

fig, ax = plt.subplots(figsize=(7, 5))

# Student measurements
for name in MY_OFFSETS:
    off, off_err = MY_OFFSETS[name]
    age, age_err = AGES_KA[name]
    ax.errorbar(age, off, xerr=age_err, yerr=off_err,
                fmt="o", color=COLS[name], ms=9, capsize=5, label=name)

# 1st abandoned — literature values only
ax.errorbar(19.3, 475, xerr=1.0, yerr=25,
            fmt="s", color="0.5", ms=9, capsize=5,
            label="1st abandoned (Sieh & Jahns 1984)")

# Best-fit line through the origin weighted by age
ages_yr = np.array([AGES_KA[n][0] for n in MY_OFFSETS]) * 1e3
offs_m  = np.array([MY_OFFSETS[n][0] for n in MY_OFFSETS])
rate_fit = np.sum(offs_m * ages_yr) / np.sum(ages_yr**2)

t = np.linspace(0, 22, 100)
ax.plot(t, rate_fit * t * 1e3 / 1e3, "k--", lw=1.5,
        label=f"Best fit (your data): {rate_fit*1e3:.1f} mm/yr")
ax.plot(t, 33.9e-3 * t * 1e3, color="0.6", lw=1.5, ls=":",
        label="Sieh & Jahns (1984): 33.9 mm/yr")

ax.set_xlabel("Age (ka)", fontsize=11)
ax.set_ylabel("Offset (m)", fontsize=11)
ax.set_title("Offset vs. age — Wallace Creek, SAF", fontsize=12)
ax.legend(fontsize=9, frameon=False)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 22); ax.set_ylim(0, None)
plt.tight_layout()
plt.show()
```

## Summary

- The San Andreas fault at Wallace Creek slips at approximately **34 mm/yr** based on two measurable offset channels spanning 3.7–13 ka.
- Fault offsets are measured by identifying channel positions in fault-parallel elevation profiles on each side of the fault.
- The dominant uncertainties are in the radiocarbon ages and in the identification of channel positions from noisy topographic profiles.
- Simple error propagation combines offset and age uncertainties to give a slip rate uncertainty.
- The offset–age relationship is approximately linear, consistent with a roughly constant slip rate over the Holocene.
